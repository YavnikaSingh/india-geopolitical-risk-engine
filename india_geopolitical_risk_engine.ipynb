```python
# INDIA GEOPOLITICAL RISK & MARKET EXPOSURE ENGINE
# Quantitative Analysis
# Author: Yavnika Singh

import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. RISK SCORES
# ---------------------------------------------------------

risk_scores = pd.DataFrame({
    "Risk ID": ["R01", "R02", "R03", "R04", "R05"],
    "Risk": [
        "India-US trade tensions",
        "India-China strategic competition",
        "West Asia conflict / energy disruption",
        "Russia sanctions and energy exposure",
        "Global shipping and supply-chain disruption"
    ],
    "Risk Score": [1.00, 2.60, 4.00, 4.30, 2.95]
})

print("QUANTITATIVE RISK SCORES")
display(risk_scores)


# ---------------------------------------------------------
# 2. RISK LEVEL CLASSIFICATION
# ---------------------------------------------------------

def classify_risk(score):
    if score < 1.5:
        return "Low"
    elif score < 2.5:
        return "Low-Moderate"
    elif score < 3.5:
        return "Moderate"
    elif score < 4.5:
        return "High"
    else:
        return "Severe"

risk_scores["Risk Level"] = risk_scores["Risk Score"].apply(classify_risk)

print("\nRISK LEVELS")
display(risk_scores)


# ---------------------------------------------------------
# 3. INDICATOR CHANGES
# ---------------------------------------------------------

indicator_changes = pd.DataFrame({
    "Indicator": [
        "India-US Merchandise Exports",
        "Indian Manufacturing PMI",
        "India-China Merchandise Imports",
        "India-China Trade Deficit",
        "Brent Crude Oil Price",
        "India CPI Inflation",
        "USD/INR",
        "Indian Crude Oil Import Volume",
        "Indian Shipping/Freight Cost",
        "Indian Engineering Exports"
    ],
    "Change %": [
        14.90,
        5.57,
        7.46,
        7.53,
        71.43,
        75.91,
        4.88,
        20.80,
        12.00,
        8.40
    ]
})

print("\nINDICATOR CHANGES")
display(indicator_changes)


# ---------------------------------------------------------
# 4. CHART — RISK SCORE
# ---------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    risk_scores["Risk ID"],
    risk_scores["Risk Score"]
)

plt.title("India Geopolitical Risk Engine — Quantitative Risk Scores")
plt.xlabel("Risk ID")
plt.ylabel("Risk Score (1–5)")
plt.ylim(0, 5)

plt.show()


# ---------------------------------------------------------
# 5. CHART — INDICATOR CHANGES
# ---------------------------------------------------------

plt.figure(figsize=(11, 6))

plt.barh(
    indicator_changes["Indicator"],
    indicator_changes["Change %"]
)

plt.title("Selected Economic & Market Indicator Changes")
plt.xlabel("Change (%)")

plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 6. RANK RISKS
# ---------------------------------------------------------

ranked_risks = risk_scores.sort_values(
    "Risk Score",
    ascending=False
).reset_index(drop=True)

ranked_risks["Rank"] = ranked_risks.index + 1

print("\nRANKED RISK PROFILE")
display(
    ranked_risks[
        ["Rank", "Risk ID", "Risk", "Risk Score", "Risk Level"]
    ]
)


# ---------------------------------------------------------
# 7. ANALYST FINDINGS
# ---------------------------------------------------------

print("\nKEY ANALYST FINDINGS")
print("-" * 60)

print("""
1. Energy-related geopolitical exposure is the strongest signal
   in the current model, with R04 and R03 recording the highest
   quantitative risk scores.

2. R04 — Russia sanctions and energy exposure — records the
   highest score at 4.30, reflecting the combined pressure from
   oil prices, crude import exposure, inflation and exchange-rate
   indicators.

3. R03 — West Asia conflict / energy disruption — scores 4.00.
   The model captures the potential transmission of energy
   disruption through crude prices, inflation and the rupee.

4. R05 — Global shipping and supply-chain disruption — scores
   2.95, indicating moderate exposure in the current model.

5. R02 — India-China strategic competition — scores 2.60.
   The model highlights trade and supply-chain exposure rather
   than treating geopolitical competition as a purely political
   issue.

6. R01 — India-US trade tensions — records the lowest quantitative
   score at 1.00 in this particular indicator sample. This should
   not be interpreted as meaning the geopolitical issue itself
   is unimportant; it reflects the indicators and weights used
   in this experimental model.
""")


# ---------------------------------------------------------
# 8. EXPORT RESULTS
# ---------------------------------------------------------

risk_scores.to_csv(
    "quantitative_risk_scores.csv",
    index=False
)

indicator_changes.to_csv(
    "indicator_changes.csv",
    index=False
)

ranked_risks.to_csv(
    "ranked_risk_profile.csv",
    index=False
)

print("\nAnalysis complete.")
print("Three CSV files have been created for your portfolio.")
```

```python
# ---------------------------------------------------------
# SCENARIO & SENSITIVITY ANALYSIS
# ---------------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt

# Scenario scores from the project's Scenario Engine
scenario_scores = pd.DataFrame({
    "Risk ID": ["R01","R01","R01",
                "R02","R02","R02",
                "R03","R03","R03",
                "R04","R04","R04",
                "R05","R05","R05"],

    "Risk": [
        "India-US trade tensions",
        "India-US trade tensions",
        "India-US trade tensions",

        "India-China strategic competition",
        "India-China strategic competition",
        "India-China strategic competition",

        "West Asia conflict / energy disruption",
        "West Asia conflict / energy disruption",
        "West Asia conflict / energy disruption",

        "Russia sanctions and energy exposure",
        "Russia sanctions and energy exposure",
        "Russia sanctions and energy exposure",

        "Global shipping and supply-chain disruption",
        "Global shipping and supply-chain disruption",
        "Global shipping and supply-chain disruption"
    ],

    "Scenario": [
        "Baseline","Escalation","De-escalation",
        "Baseline","Escalation","De-escalation",
        "Baseline","Escalation","De-escalation",
        "Baseline","Escalation","De-escalation",
        "Baseline","Escalation","De-escalation"
    ],

    "Scenario Score": [
        2,4,1,
        3,5,2,
        4,5,2,
        3,5,2,
        3,5,2
    ]
})

print("SCENARIO SCORE MATRIX")
display(scenario_scores)


# ---------------------------------------------------------
# SCENARIO MATRIX
# ---------------------------------------------------------

scenario_matrix = scenario_scores.pivot(
    index="Risk ID",
    columns="Scenario",
    values="Scenario Score"
)

scenario_matrix = scenario_matrix[
    ["Baseline","Escalation","De-escalation"]
]

print("SCENARIO MATRIX")
display(scenario_matrix)


# ---------------------------------------------------------
# ESCALATION PRESSURE
# ---------------------------------------------------------

scenario_matrix["Escalation Delta"] = (
    scenario_matrix["Escalation"]
    - scenario_matrix["Baseline"]
)

print("ESCALATION PRESSURE")
display(scenario_matrix)


# ---------------------------------------------------------
# SCENARIO CHART
# ---------------------------------------------------------

scenario_matrix[
    ["Baseline","Escalation","De-escalation"]
].plot(
    kind="bar",
    figsize=(11,6)
)

plt.title("Geopolitical Risk Scenario Analysis")
plt.xlabel("Risk ID")
plt.ylabel("Scenario Severity Score (1–5)")
plt.ylim(0,5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# SENSITIVITY ANALYSIS
# ---------------------------------------------------------

sensitivity = risk_scores.copy()

sensitivity["Escalation Score"] = [
    4,5,5,5,5
]

sensitivity["Score Increase"] = (
    sensitivity["Escalation Score"]
    - sensitivity["Risk Score"]
)

sensitivity["Relative Increase %"] = (
    sensitivity["Score Increase"]
    / sensitivity["Risk Score"]
) * 100

print("SENSITIVITY TO ESCALATION")
display(
    sensitivity[
        [
            "Risk ID",
            "Risk",
            "Risk Score",
            "Escalation Score",
            "Score Increase",
            "Relative Increase %"
        ]
    ]
)


# ---------------------------------------------------------
# RISK HEATMAP-STYLE TABLE
# ---------------------------------------------------------

risk_heatmap = scenario_matrix[
    ["Baseline","Escalation","De-escalation"]
].copy()

print("RISK SCENARIO HEATMAP")
display(
    risk_heatmap.style
    .background_gradient(cmap="RdYlGn_r", vmin=1, vmax=5)
)


# ---------------------------------------------------------
# ANALYST INTERPRETATION
# ---------------------------------------------------------

print("""
ANALYST INTERPRETATION
------------------------------------------------------------

1. R04 has the highest baseline quantitative score at 4.30,
   while its escalation scenario reaches 5.

2. R03 has a baseline score of 4.00 and also reaches 5 under
   escalation, indicating substantial sensitivity to a worsening
   energy-disruption scenario.

3. R05 and R02 have moderate baseline scores but both rise to
   5 under the escalation scenario.

4. R01 has the lowest quantitative score in the current model,
   but its escalation scenario rises from 2 to 4 in the scenario
   framework.

5. The scenario analysis demonstrates why a geopolitical risk
   assessment should not rely only on the current-state score.
   The potential change in conditions is also important.

6. The model is an experimental analytical framework. Scenario
   scores represent analyst-defined severity assumptions and are
   not probabilities or forecasts.
""")


# ---------------------------------------------------------
# EXPORT SCENARIO RESULTS
# ---------------------------------------------------------

scenario_scores.to_csv(
    "scenario_scores.csv",
    index=False
)

scenario_matrix.to_csv(
    "scenario_matrix.csv"
)

sensitivity.to_csv(
    "risk_sensitivity_analysis.csv",
    index=False
)

print("Scenario analysis complete.")
```

```python
print("""
INDIA GEOPOLITICAL RISK & MARKET EXPOSURE ENGINE
================================================

PURPOSE
-------
This model examines how geopolitical risks may transmit into
India's economy through trade, energy, financial and supply-chain
channels.

ANALYTICAL FRAMEWORK
--------------------
Geopolitical Event
        |
        v
Transmission Channel
        |
        v
Economic / Market Indicator
        |
        v
Sector Exposure
        |
        v
Quantitative Risk Score
        |
        v
Scenario Analysis

QUANTITATIVE MODEL
------------------
Each risk is evaluated using multiple indicators.

1. Indicator Change
   Measures movement between the baseline and latest observation.

2. Pressure Score
   Converts indicator movement into a 1–5 pressure scale.

3. Indicator Weight
   Assigns relative importance to each indicator.

4. Weighted Pressure
   Pressure Score multiplied by Indicator Weight.

5. Quantitative Risk Score
   Sum of weighted indicator pressures for each risk.

SCENARIO FRAMEWORK
------------------
Baseline       = Current conditions broadly persist
Escalation     = Geopolitical conditions deteriorate
De-escalation  = Geopolitical conditions improve

IMPORTANT LIMITATIONS
---------------------
This is an experimental analytical framework.

The model:
- is not a probability model;
- does not forecast geopolitical events;
- does not establish causal relationships;
- uses analyst-defined weights and scenario scores;
- is sensitive to indicator selection and data quality;
- should be interpreted alongside qualitative research.

The quantitative scores represent the structure of this
model, not an objective measurement of geopolitical risk.
""")
```

```python
# =========================================================
# INDIA GEOPOLITICAL RISK & MARKET EXPOSURE ENGINE
# SQL DATABASE
# =========================================================

import sqlite3
import pandas as pd

# Create SQLite database
conn = sqlite3.connect("india_geopolitical_risk.db")

# ---------------------------------------------------------
# 1. CREATE TABLES
# ---------------------------------------------------------

conn.executescript("""
DROP TABLE IF EXISTS risks;
DROP TABLE IF EXISTS indicators;
DROP TABLE IF EXISTS risk_indicators;
DROP TABLE IF EXISTS scenarios;

CREATE TABLE risks (
    risk_id TEXT PRIMARY KEY,
    risk TEXT,
    risk_channel TEXT,
    quantitative_score REAL
);

CREATE TABLE indicators (
    indicator_id TEXT PRIMARY KEY,
    indicator TEXT,
    baseline_value REAL,
    latest_value REAL,
    change_percent REAL
);

CREATE TABLE risk_indicators (
    risk_id TEXT,
    indicator_id TEXT,
    pressure_score INTEGER,
    indicator_weight REAL,
    weighted_pressure REAL
);

CREATE TABLE scenarios (
    risk_id TEXT,
    scenario TEXT,
    time_horizon TEXT,
    scenario_score INTEGER
);
""")

# ---------------------------------------------------------
# 2. INSERT RISKS
# ---------------------------------------------------------

risks = [
    ("R01", "India-US trade tensions", "Trade / exports", 1.00),
    ("R02", "India-China strategic competition", "Trade / supply chain", 2.60),
    ("R03", "West Asia conflict / energy disruption", "Energy / commodity prices", 4.00),
    ("R04", "Russia sanctions and energy exposure", "Energy / oil imports", 4.30),
    ("R05", "Global shipping and supply-chain disruption", "Shipping / freight costs", 2.95)
]

conn.executemany(
    "INSERT INTO risks VALUES (?, ?, ?, ?)",
    risks
)

# ---------------------------------------------------------
# 3. INSERT INDICATORS
# ---------------------------------------------------------

indicators = [
    ("I01", "Brent Crude Oil Price", 66.50, 114.00, 71.43),
    ("I02", "Indian Crude Oil Import Volume", 18.75, 22.65, 20.80),
    ("I03", "India CPI Inflation", 2.74, 4.82, 75.91),
    ("I04", "USD/INR", 90.01, 94.40, 4.88),
    ("I05", "India-US Merchandise Exports", 6.98, 8.02, 14.90),
    ("I06", "India-China Merchandise Imports", 11.382, 12.231, 7.46),
    ("I07", "India-China Trade Deficit", 10.62, 11.42, 7.53),
    ("I08", "Indian Shipping/Freight Cost", 100.00, 112.00, 12.00),
    ("I09", "Indian Engineering Exports", 10.72, 11.62, 8.40),
    ("I10", "Indian Manufacturing PMI", 53.90, 56.90, 5.57)
]

conn.executemany(
    "INSERT INTO indicators VALUES (?, ?, ?, ?, ?)",
    indicators
)

# ---------------------------------------------------------
# 4. INSERT RISK-INDICATOR RELATIONSHIPS
# ---------------------------------------------------------

risk_indicators = [
    ("R01","I05",1,0.70,0.70),
    ("R01","I10",1,0.30,0.30),

    ("R02","I06",3,0.40,1.20),
    ("R02","I07",3,0.40,1.20),
    ("R02","I10",1,0.20,0.20),

    ("R03","I01",5,0.40,2.00),
    ("R03","I03",5,0.30,1.50),
    ("R03","I04",2,0.20,0.40),
    ("R03","I10",1,0.10,0.10),

    ("R04","I01",5,0.40,2.00),
    ("R04","I02",4,0.25,1.00),
    ("R04","I03",5,0.20,1.00),
    ("R04","I04",2,0.15,0.30),

    ("R05","I03",5,0.30,1.50),
    ("R05","I08",4,0.25,1.00),
    ("R05","I09",1,0.25,0.25),
    ("R05","I10",1,0.20,0.20)
]

conn.executemany(
    "INSERT INTO risk_indicators VALUES (?, ?, ?, ?, ?)",
    risk_indicators
)

# ---------------------------------------------------------
# 5. INSERT SCENARIOS
# ---------------------------------------------------------

scenarios = [
    ("R01","Baseline","3-6 months",2),
    ("R01","Escalation","3-6 months",4),
    ("R01","De-escalation","3-6 months",1),

    ("R02","Baseline","6-12 months",3),
    ("R02","Escalation","6-12 months",5),
    ("R02","De-escalation","6-12 months",2),

    ("R03","Baseline","3-6 months",4),
    ("R03","Escalation","3-6 months",5),
    ("R03","De-escalation","3-6 months",2),

    ("R04","Baseline","6-12 months",3),
    ("R04","Escalation","6-12 months",5),
    ("R04","De-escalation","6-12 months",2),

    ("R05","Baseline","3-6 months",3),
    ("R05","Escalation","3-6 months",5),
    ("R05","De-escalation","3-6 months",2)
]

conn.executemany(
    "INSERT INTO scenarios VALUES (?, ?, ?, ?)",
    scenarios
)

conn.commit()

print("SQL DATABASE CREATED SUCCESSFULLY")
```

```python
# =========================================================
# SQL ANALYST QUERIES
# =========================================================

# QUERY 1 — Rank geopolitical risks
query1 = """
SELECT
    risk_id,
    risk,
    quantitative_score
FROM risks
ORDER BY quantitative_score DESC;
"""

print("QUERY 1 — RISK RANKING")
display(pd.read_sql_query(query1, conn))


# ---------------------------------------------------------
# QUERY 2 — Find the indicators creating the most pressure
# ---------------------------------------------------------

query2 = """
SELECT
    ri.risk_id,
    r.risk,
    i.indicator,
    ri.pressure_score,
    ri.indicator_weight,
    ri.weighted_pressure
FROM risk_indicators ri
JOIN risks r
    ON ri.risk_id = r.risk_id
JOIN indicators i
    ON ri.indicator_id = i.indicator_id
ORDER BY ri.weighted_pressure DESC;
"""

print("QUERY 2 — HIGHEST PRESSURE INDICATORS")
display(pd.read_sql_query(query2, conn))


# ---------------------------------------------------------
# QUERY 3 — Risks with multiple economic indicators
# ---------------------------------------------------------

query3 = """
SELECT
    r.risk_id,
    r.risk,
    COUNT(ri.indicator_id) AS indicator_count
FROM risks r
JOIN risk_indicators ri
    ON r.risk_id = ri.risk_id
GROUP BY r.risk_id, r.risk
ORDER BY indicator_count DESC;
"""

print("QUERY 3 — INDICATOR EXPOSURE BY RISK")
display(pd.read_sql_query(query3, conn))


# ---------------------------------------------------------
# QUERY 4 — Baseline vs escalation
# ---------------------------------------------------------

query4 = """
SELECT
    r.risk_id,
    r.risk,
    MAX(CASE
        WHEN s.scenario = 'Baseline'
        THEN s.scenario_score
    END) AS baseline_score,

    MAX(CASE
        WHEN s.scenario = 'Escalation'
        THEN s.scenario_score
    END) AS escalation_score,

    MAX(CASE
        WHEN s.scenario = 'Escalation'
        THEN s.scenario_score
    END)
    -
    MAX(CASE
        WHEN s.scenario = 'Baseline'
        THEN s.scenario_score
    END) AS escalation_change

FROM scenarios s
JOIN risks r
    ON s.risk_id = r.risk_id

GROUP BY r.risk_id, r.risk

ORDER BY escalation_change DESC;
"""

print("QUERY 4 — BASELINE VS ESCALATION")
display(pd.read_sql_query(query4, conn))


# ---------------------------------------------------------
# QUERY 5 — Most exposed indicators
# ---------------------------------------------------------

query5 = """
SELECT
    i.indicator_id,
    i.indicator,
    i.change_percent,
    COUNT(ri.risk_id) AS risks_linked
FROM indicators i
JOIN risk_indicators ri
    ON i.indicator_id = ri.indicator_id
GROUP BY
    i.indicator_id,
    i.indicator,
    i.change_percent
ORDER BY risks_linked DESC;
"""

print("QUERY 5 — INDICATORS LINKED TO MULTIPLE RISKS")
display(pd.read_sql_query(query5, conn))

print("SQL ANALYSIS COMPLETE")
```
